```python
import pandas as pd
```

```python
url_one= "https://raw.githubusercontent.com/idanm-commiting/Sanitation-Project/main/Food_Inspections_part_1.zip"
url_two="https://raw.githubusercontent.com/idanm-commiting/Sanitation-Project/main/Food_Inspections_part_2a.zip"
url_three="https://raw.githubusercontent.com/idanm-commiting/Sanitation-Project/main/Food_Inspections_part_2b.zip"
url_four="https://raw.githubusercontent.com/idanm-commiting/Sanitation-Project/main/Food_Inspections_part_3.zip"
DataFrame12= pd.read_csv(url_one, compression = "zip")
print("Column A contains the inspection ID numbers. Each ID is unique, so no pattern is apparent.")
print("Column B contains the DBA. Each DBA is unique, so no pattern is apparent.")
print("Column C contains the AKA name. Each AKA name is unique, so no pattern is apparent.")

import pandas as pd
import matplotlib.pyplot as plt
DataFrame12 = pd.concat([pd.read_csv(url, compression="zip", low_memory=False) for url in [url_one, url_two, url_three, url_four]], ignore_index=True)
counts = DataFrame12["Facility Type"].fillna("Blank").str.strip().replace("", "Blank").value_counts()
counts[counts >= 1000].plot.pie(autopct="%1.1f%%", figsize=(10, 10), title="Distribution of Facility Types", ylabel="")
plt.show()


DataFrame12 = pd.concat([pd.read_csv(url, compression="zip", low_memory=False) for url in [url_one, url_two, url_three, url_four]], ignore_index=True)
ax = DataFrame12.loc[DataFrame12["Facility Type"].str.strip().str.lower() == "restaurant", "Risk"].value_counts().plot.bar(title="Distribution of Risk Type for Facilities",
    xlabel="Risk Type", ylabel="Count", rot=0)
ax.bar_label(ax.containers[0], padding=3);


import pandas as pd
import matplotlib.pyplot as plt
DataFrame12 = pd.concat([pd.read_csv(url, compression="zip", low_memory=False) for url in [url_one, url_two, url_three, url_four]], ignore_index=True)
counts = DataFrame12["Zip"].dropna().astype(str).str.replace(r"\.0$", "", regex=True).value_counts().sort_index()
counts = counts[counts >= 1500]
plt.figure(figsize=(14, 5))
plt.scatter(counts.index, counts.values)
plt.title("Inspection Count by ZIP Code")
plt.xlabel("ZIP Code")
plt.ylabel("Number of Inspections")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()


print("Column D contains the Liscence number. Each the Liscence number is unique, so no pattern is apparent.")
print("Column G contains the address. Each the address is unique, so no pattern is apparent.")
print("Column H contains the city. The city is the same, so no pattern is apparent.")
print("Column I contains the State. The State is the same, so no pattern is apparent.")
```

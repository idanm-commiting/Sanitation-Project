```python
import pandas as pd
import numpy as np

food_path = ("/Users/michael818/Desktop/BA780_dataCleaning/整理好的資料/Food_Inspections_2010_2026.csv")

final_311_path = ("/Users/michael818/Desktop/BA780_dataCleaning/整理好的資料/311_Sanitation_Combined_2010_2026.csv")

food = pd.read_csv(food_path)

df_311 = pd.read_csv(final_311_path)
```

```python
print("===== FOOD INSPECTIONS =====")

# Missing values
print("\nMissing Values:")
print(food.isna().sum().sort_values(ascending=False))

# Duplicate rows
print("\nDuplicate Rows:")
print(food.duplicated().sum())

# Duplicate Inspection IDs
print("\nDuplicate Inspection IDs:")
print(food["Inspection ID"].duplicated().sum())

# Results categories
print("\nResults:")
print(food["Results"].value_counts(dropna=False))

# Risk categories
print("\nRisk:")
print(food["Risk"].value_counts(dropna=False))

# Facility Type
print("\nFacility Types:")
print(food["Facility Type"].value_counts(dropna=False).head(30))

# ZIP
print("\nZIP range:")
print(food["Zip"].min(), food["Zip"].max())

# Latitude / Longitude
print("\nLatitude range:")
print(food["Latitude"].min(), food["Latitude"].max())

print("\nLongitude range:")
print(food["Longitude"].min(), food["Longitude"].max())
```

```python
print("===== 311 SANITATION =====")

# Missing values
print("\nMissing Values:")
print(df_311.isna().sum().sort_values(ascending=False))

# Duplicate rows
print("\nDuplicate Rows:")
print(df_311.duplicated().sum())

# Duplicate Request Numbers
print("\nDuplicate Request Numbers:")
print(df_311["Request Number"].duplicated().sum())

# Status categories
print("\nStatus:")
print(df_311["Status"].value_counts(dropna=False))

# Request Type
print("\nRequest Type:")
print(df_311["Request Type"].value_counts(dropna=False))

# ZIP
print("\nZIP Codes:")
print(df_311["ZIP Code"].value_counts(dropna=False).sort_index())

# Geographic ranges
print("\nWard range:")
print(df_311["Ward"].min(),df_311["Ward"].max())

print("\nPolice District range:")
print(df_311["Police District"].min(),df_311["Police District"].max())

print("\nCommunity Area range:")
print(df_311["Community Area"].min(),df_311["Community Area"].max())

print("\nLatitude range:")
print(df_311["Latitude"].min(),df_311["Latitude"].max())

print("\nLongitude range:")
print(df_311["Longitude"].min(),df_311["Longitude"].max())
```
